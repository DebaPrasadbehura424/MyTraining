```python
import pandas as pd;

data= pd.read_csv("z2.csv")

data

df= pd.DataFrame(data)

df
```

```python
df.head()
```

```python
df["ODD"]= df["ODD"] + 1;

df
```

```python
# df.drop([9],axis=0)
# df.drop([9],axis=1)

df.dropna(subset=["ODD"],axis=0)
```

```python
df["even mul4"]= df["even mul4"]/2;
df
```

```python
df.rename(columns={"even mul4":"even mul2"});
# df["even mul2"]= df["even mul2"].astype("int");
```

```python
df["even mul4"]= df["even mul4"].astype("int");
df
```

```python
#  Data Normalizaton
# Simple Features -> X new = xold/xmax
# MINMAX -> x new = x old- xmin / xmax - xmin
# Z SCORE & standard score -> x NEW = n old - mew /standard davition
```

```python
df["sumOfConstant"]= df["sumOfConstant"]/ df["sumOfConstant"].max()
df
```

```python
df["sumOfConstant"]= (df["sumOfConstant"]- df["sumOfConstant"].min()) /( df["sumOfConstant"].max()-df["sumOfConstant"].min()); 
df
```

```python
df["sumOfConstant"]= df["sumOfConstant"]- df["sumOfConstant"].mean() / df["sumOfConstant"].std(); 
df
```
